```python
from Bio import Entrez
from Bio import Medline
```

```python
Entrez.email = "markyan@creighton.edu"

handle = Entrez.esearch(
    db="pubmed",
    term='("sports injury") AND case report',
    retmax=1000
)

record = Entrez.read(handle)
pmids = record["IdList"]
```

```python
ids = ",".join(pmids[:100])

fetch = Entrez.efetch(
    db="pubmed",
    id=ids,
    rettype="abstract",
    retmode="text"
)

text = fetch.read()

print(text[:10000])
```

```python
## Trial searches:

COMMON_INJURIES = [
    "concussion",
    "hamstring strain",
    "ankle sprain",
    "knee sprain",
    "MCL sprain",
    "ACL tear",
    "Achilles tendon injury",
    "quad strain",
    "groin strain",
    "shoulder injury",
    "wrist injury",
    "back injury"
]

all_records = []

for injury in COMMON_INJURIES:

    search = Entrez.esearch(
        db="pubmed",
        term=f'"{injury}" AND athlete',
        retmax=500
    )

    pmids = Entrez.read(search)["IdList"]

    if len(pmids) == 0:
        continue

    fetch = Entrez.efetch(
        db="pubmed",
        id=",".join(pmids),
        rettype="medline",
        retmode="text"
    )

    records = Medline.parse(fetch)

    for record in records:
        all_records.append({
            "injury": injury,
            "title": record.get("TI", ""),
            "abstract": record.get("AB", "")
        })


```
